## Discrete Distribution Questions:
1. Which champion kills the most / dies the most? (highest takedowns, highest deaths)
2. Which champion makes the most money? 
3. Most common champion per lane?

```
def visualize_barplot(
    df: pd.DataFrame,
    x: str, # Champion
    y: str, # Kills
    **kwargs
) -> None:
    fig, ax = plt.subplots(figsize=(10,6))
    sns.barplot(
        data=df,
        x=x,
        y=y,
        ax=ax,
        **kwargs
    )
    
    ax.set_title(f"Top 10 {x.lower()} by {y.lower().replace("_", " ")}")
    ax.set_xlabel(x.replace("_", " ").title())
    ax.set_ylabel(y.replace("_", " ").title())
    
    ax.tick_params(axis='x', rotation=45)
    plt.setp(ax.get_xticklabels(), ha='right')
    plt.tight_layout()
    
    plt.show()
    plt.close(fig)
```

### 1. Which champion kills the most / dies the most? 

```
def the_most_champion(
    df: pd.DataFrame,
    by: str,
    top_n: int = 10,
    show_plots: bool = True,
    **vis_kwargs
) -> pd.DataFrame:
    df = (df
        .groupby("CHAMPION", as_index=False)
        .agg(**{
            f"{by}_MEAN": (by, lambda x: np.round(x.mean(), 2))
        })
        .sort_values(f"{by}_MEAN", ascending=False)
        .head(top_n)
    )

    if show_plots:
        visualize_barplot(df, x="CHAMPION", y=f"{by}_MEAN", **vis_kwargs)

    return df
```

Karthus is the champion with the most deaths overall, leading by a noticable margin. Attributable to the champion's playstyle that encourages death to activate his passive.

```
the_most_champion(
    df=player_match_final_stats.to_pandas(),
    by='DEATHS',
    top_n=5,
    show_plots=True,
    color='Salmon'
)
```

The distribution of most kills for champions is more even out, with mostly assassins occupying the top 10 (e.g. Briar, Katarina, Evelynn, Talon, Rengar, Khazix)

```
the_most_champion(
    df=player_match_final_stats.to_pandas(),
    by="KILLS",
    top_n=10,
    show_plots=True,
    color='Green'
)
```
