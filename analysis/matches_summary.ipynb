```
USE WAREHOUSE COMPUTE_WH;

USE DATABASE TEST_PIPELINE_DB;

USE SCHEMA BRONZE;
```

```
%%sql -r matches_summary
SELECT 
    MATCH_ID,
    GAME_DURATION,
    CASE 
        WHEN WINNING_TEAM = 100 THEN 'Blue'
        WHEN WINNING_TEAM = 200 THEN 'Red'
        ELSE NULL
    END AS WINNING_TEAM,
    AVERAGE_RANK
FROM MATCHES_SUMMARY_BRONZE
;
```
